```python
# Hypothesis Testing
```

```python
#create Q-Q plot with 45-degree line added to plot
sm.qqplot(df['fare_amount'], line='45')
plt.show()
```

```python
# sample 1
card_sample = df[df['payment_type'] == 'Card']['fare_amount']

# sample 2
cash_sample = df[df['payment_type'] == 'Cash']['fare_amount']

# performing t test on both the different sample
t_stat, p_value = st.ttest_ind(a=card_sample, b=cash_sample, equal_var=False)
print('T-statistic', t_stat, 'p-value', p_value)

# comparing the p value with the significance of 5% or 0.05
if p_value < 0.05:
    print("\nReject the null hypothesis")
else:
    print("\nAccept the null hypothesis")
```

```python
# A/B Testing: Payment Method Impact on Fare
# Business Problem:
# Taxi companies want to understand whether payment method affects revenue?
# Do customers who pay by card generate higher fare compared to cash users?
```

```python
# A/B Testing: Card vs Cash

# Group A → Cash
group_A = df[df['payment_type'] == 'Cash']['fare_amount']
# Group B → Card
group_B = df[df['payment_type'] == 'Card']['fare_amount']

# Mean comparison
print("Cash Avg Fare:", group_A.mean())
print("Card Avg Fare:", group_B.mean())

# T-test
from scipy import stats

t_stat, p_value = stats.ttest_ind(group_A, group_B, equal_var=False)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
alpha = 0.05

if p_value < alpha:
    print("Reject Null Hypothesis → Payment type impacts fare")
else:
    print("Fail to Reject Null → No significant impact of payment type")
```

```python
### Conclusion

The A/B test compared average fare between cash and card users.

- If p-value < 0.05 → Payment method has a significant impact on fare
- If p-value ≥ 0.05 → No significant difference observed

This analysis helps businesses understand customer behavior and optimize payment strategies.
```
